```python
# ==============================================================================
# 1. NUMPY: Wektoryzacja i Wymiary Macierzy (Shapes)
# ==============================================================================
import numpy as np

# A. Wektoryzacja vs Pętla w Pythonie
zarobki_brutto = np.array([4500, 6000, 8500, 12000, 15000])

# Zamiast pętli 'for x in zarobki: x * 0.8', robimy operację bezpośrednio:
zarobki_netto = zarobki_brutto * 0.72 - 250
print("1. Wektoryzacja płac netto:", zarobki_netto)

# B. Dlaczego wymiary (Shapes) są kluczowe w ML?
cecha_x = np.array([10, 20, 30, 40, 50])
print("\n2. Wektor 1D (Target y):", cecha_x.shape) # (5,)

# Zamiana na macierz 2D (Macierz X wymagana przez Scikit-Learn)
cecha_x_2d = cecha_x.reshape(-1, 1)
print("3. Macierz 2D po reshape(-1, 1) (Cecha X):", cecha_x_2d.shape) # (5, 1)

# C. Szybkie operacje statystyczne
print(f"Średnia: {cecha_x.mean():.2f} | Odchylenie std: {cecha_x.std():.2f}")
```

```python
# ==============================================================================
# 2. PANDAS: Tworzenie tabeli, inspekcja techniczna i biznesowa
# ==============================================================================
import pandas as pd

# Tworzymy przykładową tabelę klientów banku
dane_klientow = {
    'Klient_ID': [101, 102, 103, 104, 105, 106],
    'Wiek': [25, 42, np.nan, 35, 61, 19],          # Zawiera brakujący wiek (NaN)
    'Zarobki_PLN': [4500, 12000, 8500, 7200, 19000, 3100],
    'Segment': ['Standard', 'VIP', 'Standard', 'Standard', 'VIP', 'Junior'],
    'Decyzja_Kredyt': [1, 1, 0, 1, 1, 0]          # Target y (1 = Przyznano, 0 = Odrzucono)
}

df = pd.DataFrame(dane_klientow)

print("--- 1. Pierwsze wiersze tabeli (head) ---")
display(df.head(3))

print("\n--- 2. Podsumowanie typów i braków danych (info) ---")
df.info()

print("\n--- 3. Szybka statystyka (describe) ---")
display(df.describe().T)

print("\n--- 4. Zbalansowanie klas targetu ---")
print(df['Decyzja_Kredyt'].value_counts(normalize=True).round(2))
```

```python
# ==============================================================================
# 3. PANDAS: Filtrowanie, Imputacja i Przygotowanie pod Scikit-Learn
# ==============================================================================

# A. Filtrowanie warunkowe (np. tylko segment VIP lub zarobki > 8000)
klienci_bogaci = df[(df['Zarobki_PLN'] > 8000) & (df['Segment'] == 'VIP')]
print("1. Klienci VIP z wysokim dochodem:")
display(klienci_bogaci)

# B. Prosta obsługa braku danych (NaN) w Pandas
mediana_wieku = df['Wiek'].median()
df['Wiek_Imputowany'] = df['Wiek'].fillna(mediana_wieku)
print(f"\n2. Braki w wieku zastąpione medianą ({mediana_wieku} lat).")

# C. Święty Graal przed ML: Podział na macierz cech (X) i wektor wynikowy (y)
X = df[['Wiek_Imputowany', 'Zarobki_PLN']] # Zmienne objaśniające
y = df['Decyzja_Kredyt']                  # Zmienna objaśniana (Target)

print("\n3. Macierz X (Cechy):")
display(X)
print("4. Wektor y (Etykiety):")
print(y.values)
```
